```python
# ==================== Load data ====================
# the next two datasets are 200 400 phase windowed datasets with newer updated sta lta values

dataType = 'complete'
networkName = 'swp'
stationName = 'swp'
freqType = 'normal'

raw_swp_eventdata = pd.read_csv(
    f'../data/{dataType}/{stationName}/{networkName}_measured_events_{freqType}_04032026.txt',
    sep='|'
)

dataType = 'complete'
networkName = 'AU'
stationName = 'australia'
freqType = 'normal'

raw_aus_eventdata = pd.read_csv(
    f'../data/{dataType}/{stationName}/{networkName}_measured_events_{freqType}_04032026.txt',
    sep='|'
)

raw_eventdata = pd.concat([raw_swp_eventdata, raw_aus_eventdata], ignore_index=True)
raw_eventdata = raw_eventdata.sort_values('evstation')

# ==================== Prepare data ====================
# Calculate SI errors
eventData = filtering.calculate_si_errors(raw_eventdata)

print("length before filtering:", len(eventData))
s_count = (eventData["evphase"] == 'S').sum()
sks_count = (eventData["evphase"] == 'SKS').sum()

print("S count:", s_count)
print("SKS count:", sks_count)

print("Number of Seismic Stations Before Filtering:", eventData['evstation'].nunique())
print(eventData['evstation'].unique())
# print(eventData['evstation'].value_counts())



epicentral_max=360
epicentral_min=0
depth_min=0
depth_max=1e5
snrbefore_min=3.0
snrafter_min=1.0
sta_lta_min=0
mag_max=7.0
ellipticity_min=1.0
ellipticity_max=1e4
si_error_max=0.5
incident_max=90
pierce_depth_min=0
delta_travel_max = 25


# Apply quality filters
eventData = filtering.filter_events(
    eventData,
    epicentral_max=epicentral_max,
    epicentral_min=epicentral_min,
    depth_min=depth_min,
    depth_max=depth_max,
    snrbefore_min=snrbefore_min,
    snrafter_min=snrafter_min,
    sta_lta_min=sta_lta_min,
    mag_max=mag_max,
    ellipticity_min=ellipticity_min,
    ellipticity_max=ellipticity_max,
    si_error_max=si_error_max,
    incident_max=incident_max,
    pierce_depth_min=pierce_depth_min,
    delta_travel_max=delta_travel_max
)

eventData = eventData.drop_duplicates(subset=['evtime','evstation'])

# Remove shallow events with large incident angles (keeps only events with large epicentral distances and smaller incident angles)
eventData = eventData[~((eventData['evdp'] < 400) & (eventData['evincident'] > 25))]

# # Remove deep events with very large incident angles (sample)
eventData = eventData[~((eventData['evdp'] >= 400) & (eventData['evincident'] > 35))]

# ^^^^^^ these two filters are heavily used in the new analysis so keep them!!!


'''
april 5 notes
- added tilt to SI calculation and plotting (see analysis.py and spring_updates.ipynb)
- started looking at treating deep S phases and direct S phases separately (mainly dealing with incident angle and pierce depth filters)
- analysis of only deep S phases to identify trends and how this could be corraborated with sampling of anisotropy at tilted layer)
'''
print("length after filtering:", len(eventData))
s_count = (eventData["evphase"] == 'S').sum()
sks_count = (eventData["evphase"] == 'SKS').sum()

print("S count:", s_count)
print("SKS count:", sks_count)

print("Number of Seismic Stations After Filtering:", eventData['evstation'].nunique())


print(f"Events after filtering: {len(eventData)}")
print(eventData['evstation'].unique())

print(eventData['evstation'].value_counts().to_string())
```

```python
import pandas as pd
import numpy as np


# ============================================================
# PARAMETERS
# ============================================================

phi_bin_width = 10
n_phi_bins = 18
min_measurements = 10
s_depth = 400


# ============================================================
# 1. LOAD / COMBINE RAW DATA
# ============================================================

# These should already exist in your notebook:
# raw_swp_eventdata
# raw_aus_eventdata

raw_eventdata = pd.concat(
    [
        raw_swp_eventdata,
        raw_aus_eventdata
    ],
    ignore_index=True
)

raw_eventdata = raw_eventdata.sort_values(
    "evstation"
).reset_index(drop=True)


# ============================================================
# 2. CALCULATE SI ERRORS
# ============================================================

eventData = filtering.calculate_si_errors(
    raw_eventdata
)


# ============================================================
# 3. APPLY ESTABLISHED QC FILTERS
# ============================================================

eventData = filtering.filter_events(
    eventData,
    epicentral_max=360,
    epicentral_min=0,
    depth_min=0,
    depth_max=1e5,
    snrbefore_min=3.0,
    snrafter_min=1.0,
    sta_lta_min=0,
    mag_max=7.0,
    ellipticity_min=1.0,
    ellipticity_max=1e4,
    si_error_max=0.5,
    incident_max=90,
    pierce_depth_min=0,
    delta_travel_max=25
)


# ============================================================
# 4. REMOVE DUPLICATES
# ============================================================

eventData = eventData.drop_duplicates(
    subset=["evtime", "evstation"]
).copy()


# ============================================================
# 5. APPLY DEPTH-DEPENDENT INCIDENT ANGLE FILTER
# ============================================================

# Shallow events: evdp < 400 km
eventData = eventData[
    ~(
        (eventData["evdp"] < s_depth)
        & (eventData["evincident"] > 25)
    )
].copy()

# Deep events: evdp >= 400 km
eventData = eventData[
    ~(
        (eventData["evdp"] >= s_depth)
        & (eventData["evincident"] > 35)
    )
].copy()


# ============================================================
# 6. DEFINE RETAINED STATION POPULATION
# ============================================================

station_counts = (
    eventData
    .groupby("evstation")
    .size()
)

# At least 10 accepted measurements
station_list = station_counts[
    station_counts >= min_measurements
].index.tolist()

# Only stations represented in STATION_PHASE_MAP
station_list = [
    station
    for station in station_list
    if station in STATION_PHASE_MAP
]


# Filter eventData to final 59-station population
df = eventData[
    eventData["evstation"].isin(station_list)
].copy()


print(
    "Stations retained:",
    df["evstation"].nunique()
)

print(
    "Measurements retained:",
    len(df)
)


# ============================================================
# 7. VALIDATE STATION / MEASUREMENT POPULATION
# ============================================================

assert df["evstation"].nunique() == 59
assert len(df) == 7828


# ============================================================
# 8. CALCULATE POLARIZATION AZIMUTH
# ============================================================

# calcbaz is the measured polarization azimuth.
#
# Because first-order azimuthal anisotropy is 180-degree
# periodic, fold 0–360 degrees into 0–180 degrees.

df["phi_180"] = (
    df["calcbaz"] % 180
)


# ============================================================
# 9. ASSIGN 10-DEGREE POLARIZATION-AZIMUTH BINS
# ============================================================

df["phi_bin"] = np.floor(
    df["phi_180"] / phi_bin_width
).astype(int)

# Valid bins are 0–17 for 10-degree bins
assert df["phi_bin"].between(
    0,
    n_phi_bins - 1
).all()


# ============================================================
# 10. LOAD THE S2 TXT FILE
# ============================================================

new_measurements_path = (
    "/work/gcl3/BR/nicolasv03/senior_thesis/"
    "measurements/publication_map/"
    "swp_measurements_corrected_jul292026.txt"
)

new_data_df = pd.read_csv(
    new_measurements_path,
    sep="\t",
    encoding="cp1252"
)


# ============================================================
# 11. BUILD STATION INFORMATION FROM STATION_PHASE_MAP
# ============================================================

phase_df = pd.DataFrame(
    STATION_PHASE_MAP.items(),
    columns=[
        "Station",
        "phase_dataset"
    ]
)

phase_df["Class"] = (
    phase_df["phase_dataset"]
    .map({
        "sks": "Simple",
        "sks_deep": "Simple",
        "all": "Simple",
        "complex": "Complex"
    })
    .fillna("Unknown")
)


# ============================================================
# 12. BUILD S1
# ============================================================

s1 = (
    df
    .groupby("evstation")
    .size()
    .rename("Total")
    .to_frame()
)

# ------------------------------------------------------------
# SKS counts
# ------------------------------------------------------------

s1["SKS"] = (
    df["evphase"]
    .eq("SKS")
    .groupby(df["evstation"])
    .sum()
)

# ------------------------------------------------------------
# Deep S counts
# ------------------------------------------------------------

s1["Deep S"] = (
    (
        (df["evphase"] == "S")
        &
        (df["evdp"] >= s_depth)
    )
    .groupby(df["evstation"])
    .sum()
)

# ------------------------------------------------------------
# Shallow S counts
# ------------------------------------------------------------

s1["Shallow S"] = (
    (
        (df["evphase"] == "S")
        &
        (df["evdp"] < s_depth)
    )
    .groupby(df["evstation"])
    .sum()
)

# ------------------------------------------------------------
# Convert index into Station column
# ------------------------------------------------------------

s1 = (
    s1
    .reset_index()
    .rename(columns={"evstation": "Station"})
)

# ------------------------------------------------------------
# Add station network and coordinates
# ------------------------------------------------------------

station_metadata = (
    df.groupby("evstation")
    .agg(
        Network=("evnetwork", lambda x: ", ".join(sorted(x.dropna().unique()))),
        Latitude=("evstationlat", "first"),
        Longitude=("evstationlon", "first"),
    )
    .reset_index()
    .rename(columns={"evstation": "Station"})
)

s1 = s1.merge(
    station_metadata,
    on="Station",
    how="left"
)

s1["Latitude"] = s1["Latitude"].map(
    lambda x: f"{x:.3f}" if pd.notna(x) else ""
)

s1["Longitude"] = s1["Longitude"].map(
    lambda x: f"{x:.3f}" if pd.notna(x) else ""
)


# ============================================================
# 13. ADD STATION CLASSIFICATION / PHASE CONFIGURATION
# ============================================================

s1["phase_dataset"] = (
    s1["Station"]
    .map(STATION_PHASE_MAP)
)

s1["Class"] = (
    s1["phase_dataset"]
    .map({
        "sks": "Simple",
        "sks_deep": "Simple",
        "all": "Simple",
        "complex": "Complex"
    })
)


# ============================================================
# 14. CALCULATE POLARIZATION-AZIMUTH COVERAGE
# ============================================================

phi_coverage = []


for station, group in df.groupby("evstation"):

    phase_dataset = STATION_PHASE_MAP[station]


    # --------------------------------------------------------
    # SKS-only polarization-azimuth coverage
    # --------------------------------------------------------

    sks_group = group[
        group["evphase"] == "SKS"
    ]

    sks_bins = (
        sks_group["phi_bin"]
        .nunique()
    )


    # --------------------------------------------------------
    # Simple stations
    # --------------------------------------------------------

    if phase_dataset == "sks":

        # Analysis used SKS only
        analysis_group = group[
            group["evphase"] == "SKS"
        ]

        analysis_bins = (
            analysis_group["phi_bin"]
            .nunique()
        )

        delta_bins = int(
            analysis_bins
            - sks_bins
        )


    elif phase_dataset == "sks_deep":

        # Analysis used SKS + Deep S
        analysis_group = group[
            (
                group["evphase"] == "SKS"
            )
            |
            (
                (group["evphase"] == "S")
                &
                (group["evdp"] >= s_depth)
            )
        ]

        analysis_bins = (
            analysis_group["phi_bin"]
            .nunique()
        )

        delta_bins = int(
            analysis_bins
            - sks_bins
        )


    elif phase_dataset == "all":

        # Analysis used SKS + Deep S + Shallow S
        analysis_group = group.copy()

        analysis_bins = (
            analysis_group["phi_bin"]
            .nunique()
        )

        delta_bins = int(
            analysis_bins
            - sks_bins
        )


    # --------------------------------------------------------
    # Complex stations
    #
    # Exact phase configuration was not preserved.
    #
    # Therefore:
    #   SKS_phi_bins      = SKS observations
    #   Analysis_phi_bins = all observed usable measurements
    #   Delta_phi_bins    = not applicable
    # --------------------------------------------------------

    elif phase_dataset == "complex":

        analysis_group = group.copy()

        analysis_bins = (
            analysis_group["phi_bin"]
            .nunique()
        )

        delta_bins = np.nan


    else:

        analysis_bins = np.nan
        delta_bins = np.nan


    phi_coverage.append(
        {
            "Station": station,
            "SKS_phi_bins": sks_bins,
            "Analysis_phi_bins": analysis_bins,
            "Delta_phi_bins": delta_bins
        }
    )


phi_coverage_df = pd.DataFrame(
    phi_coverage
)


# ============================================================
# 15. MERGE POLARIZATION COVERAGE INTO S1
# ============================================================

s1 = s1.merge(
    phi_coverage_df,
    on="Station",
    how="left"
)


# ============================================================
# 16. FINAL S1 COLUMN ORDER
# ============================================================

s1 = s1[
    [
        "Station",
        "Network",
        "Latitude",
        "Longitude",
        "Class",
        "Total",
        "SKS",
        "Deep S",
        "Shallow S",
        "SKS_phi_bins",
        "Analysis_phi_bins",
        "Delta_phi_bins"
    ]
]


# ============================================================
# 17. SORT S1
# ============================================================

s1 = (
    s1
    .sort_values("Station")
    .reset_index(drop=True)
)


# ============================================================
# 18. S1 VALIDATION
# ============================================================

print(
    "\n========== S1 VALIDATION =========="
)

print(
    "Stations:",
    len(s1)
)

print(
    "Measurements:",
    s1["Total"].sum()
)

print(
    "SKS:",
    s1["SKS"].sum()
)

print(
    "Deep S:",
    s1["Deep S"].sum()
)

print(
    "Shallow S:",
    s1["Shallow S"].sum()
)


# Station count
assert len(s1) == 59

# Total measurement count
assert (
    s1["Total"].sum()
    == 7828
)

# Phase counts
assert (
    s1["SKS"].sum()
    == 838
)

assert (
    s1["Deep S"].sum()
    == 2227
)

assert (
    s1["Shallow S"].sum()
    == 4763
)

# Phase counts must sum to total
assert (
    s1["SKS"].sum()
    +
    s1["Deep S"].sum()
    +
    s1["Shallow S"].sum()
    ==
    7828
)

s1["SKS_phi_bins"] = s1["SKS_phi_bins"].astype("Int64")
s1["Analysis_phi_bins"] = s1["Analysis_phi_bins"].astype("Int64")
s1["Delta_phi_bins"] = s1["Delta_phi_bins"].astype("Int64")

# ============================================================
# 19. ADD S1 TOTAL ROW
# ============================================================

s1_total = pd.DataFrame(
    [
        {
            "Station": "Total",
            "Network": "",
            "Latitude": "",
            "Longitude": "",
            "Class": "",
            "Total": s1["Total"].sum(),
            "SKS": s1["SKS"].sum(),
            "Deep S": s1["Deep S"].sum(),
            "Shallow S": s1["Shallow S"].sum(),
            "SKS_phi_bins": "",
            "Analysis_phi_bins": "",
            "Delta_phi_bins": ""
        }
    ]
)


s1 = pd.concat(
    [
        s1,
        s1_total
    ],
    ignore_index=True
)


# ============================================================
# 20. FORMAT S1 FOR LATEX
# ============================================================

print(
    "\n========== TABLE S1 =========="
)

print(
    s1.to_latex(
        index=False,
        escape=False
    )
)

# export S1 to excel sheet
# s1.to_excel('')

s1.to_excel("/work/gcl3/BR/nicolasv03/senior_thesis/measurements/supp_info_tables/S1.xlsx", index=False)


# ============================================================
# 21. BUILD MASTER STATION DATAFRAME FOR S2
#
# The TXT file is authoritative for the final station-level
# splitting parameters and phase counts.
# ============================================================

station_df = phase_df.merge(
    new_data_df,
    on="Station",
    how="left"
)


# ============================================================
# 22. STATION CLASSIFICATION
# ============================================================

station_df["classification"] = (
    station_df["phase_dataset"]
    .map({
        "sks": "Simple",
        "sks_deep": "Simple",
        "all": "Simple",
        "complex": "Complex"
    })
    .fillna("Unknown")
)


# ============================================================
# 23. KEEP ONLY PHASES ACTUALLY USED IN FINAL MEASUREMENT
#
# This preserves the original S2 logic.
# ============================================================

station_df.loc[
    station_df.phase_dataset == "sks",
    [
        "s_deep_count",
        "s_shallow_count"
    ]
] = 0


station_df.loc[
    station_df.phase_dataset == "sks_deep",
    "s_shallow_count"
] = 0


station_df["total_count"] = (
    station_df["sks_count"]
    +
    station_df["s_deep_count"]
    +
    station_df["s_shallow_count"]
)


# ============================================================
# 24. TABLE S2
#
# Only interpreted/simple stations.
# ============================================================

results_table = (
    station_df.loc[
        station_df["classification"] == "Simple",
        [
            "Station",
            "phase_dataset",
            "phi",
            "dt"
        ]
    ]
    .copy()
)


results_table = (
    results_table
    .rename(
        columns={
            "phase_dataset": "Phases",
            "phi": r"$\phi$",
            "dt": r"$\delta t$"
        }
    )
)


results_table["Phases"] = (
    results_table["Phases"]
    .replace(
        {
            "sks": "SKS",
            "sks_deep": "SKS + Deep S",
            "all": "SKS + Deep S + Shallow S"
        }
    )
)


for col in [
    r"$\phi$",
    r"$\delta t$"
]:

    results_table[col] = (
        results_table[col]
        .map(
            lambda x:
                f"{x:g}"
                if pd.notna(x)
                else "-"
        )
    )


# ============================================================
# 25. S2 VALIDATION / COUNT
# ============================================================

print(
    "\n========== S2 VALIDATION =========="
)

print(
    "Simple stations in S2:",
    len(results_table)
)

print(
    "Complex stations:",
    (
        station_df["classification"]
        .eq("Complex")
        .sum()
    )
)

# This is intentionally printed rather than asserted against 24,
# because your current STATION_PHASE_MAP / TXT may contain 26
# stations classified as Simple even though the manuscript text
# currently says 24 first-order stations.

if len(results_table) != 24:

    print(
        "\nWARNING:"
        "\nS2 currently contains "
        f"{len(results_table)} simple stations, "
        "while the manuscript currently states 24."
        "\nDo not force this to 24 in code; "
        "reconcile the two station populations separately."
    )


# ============================================================
# 26. S2 SUPPLEMENTARY CATALOG TABLE
#
# Every processed station represented in the TXT.
# ============================================================

catalog_table = station_df[
    [
        "Station",
        "classification",
        "phase_dataset",
        "total_count",
        "sks_count",
        "s_deep_count",
        "s_shallow_count"
    ]
].copy()


catalog_table = (
    catalog_table
    .rename(
        columns={
            "classification": "Class",
            "phase_dataset": "Phases",
            "total_count": "Total",
            "sks_count": "SKS",
            "s_deep_count": "Deep S",
            "s_shallow_count": "Shallow S"
        }
    )
)


catalog_table["Phases"] = (
    catalog_table["Phases"]
    .replace(
        {
            "sks": "SKS",
            "sks_deep": "SKS + Deep S",
            "all": "All",
            "complex": "-"
        }
    )
)


catalog_table = (
    catalog_table
    .fillna("-")
)


for col in [
    "Total",
    "SKS",
    "Deep S",
    "Shallow S"
]:

    catalog_table[col] = (
        catalog_table[col]
        .map(
            lambda x:
                f"{x:g}"
                if isinstance(
                    x,
                    (int, float, np.floating)
                )
                else x
        )
    )


# ============================================================
# 27. PRINT S2 TABLES
# ============================================================

print(
    "\n========== TABLE S2 =========="
)

print(
    results_table.to_latex(
        index=False,
        escape=False
    )
)


print(
    "\n========== S2 CATALOG TABLE =========="
)

print(
    catalog_table.to_latex(
        index=False,
        escape=False
    )
)
```
